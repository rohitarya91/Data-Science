```python
import pandas as pd 
import numpy as np

dataset = pd.read_csv("dirty_cafe_sales.csv", na_values=["?", " ", "????", "??", "???", "UNKNOWN" , "ERROR"])

# dataset.head()



# dataset.columns
dataset['Item']
dataset['Item'] = dataset['Item'].fillna(dataset['Item'].mode()[0]) # replaceing missing value as per mode 
dataset['Quantity'] = dataset['Quantity'].fillna(dataset['Quantity'].mean().round(0)) #replaceing missing value as per mean 
dataset['Price Per Unit'] = dataset['Price Per Unit'].fillna(dataset['Price Per Unit'].mean().round(0)) #replaceing missing value as per mean 
dataset['Total Spent'] = dataset['Total Spent'].fillna(dataset['Total Spent'].mean().round(0)) #replaceing missing value as per mean 
pay_method_map = { "Credit Card" : 0 , "Cash" : 1 , "Digital Wallet" : 2 , } # creating a dict for reaplceing character value bainary value 

dataset['Payment Method'] = dataset['Payment Method'].map(pay_method_map) # replaceing value with help of map function 
dataset['Payment Method'] = dataset['Payment Method'].fillna(dataset['Payment Method'].mean().round(0)) # fill missing value as per mean 

# now creating a dict for location variabale and canvert values character to binary 
location_map = { "In-store" : 0 , "Takeaway" : 1 }

#now replaceing value 

dataset['Location'] = dataset['Location'].map(location_map)
dataset['Location'] = dataset['Location'].fillna(dataset['Location'].mean().round())
dataset['Location'].unique()
dataset.isnull().sum()
```
